Программа ищет фильм в OMDb, получает JSON и извлекает отдельные поля.
Для запуска нужен личный OMDb API key. Ячейки выполняются сверху вниз.

```python
import json
import getpass
import requests
```

Ключ читается из локального `.env`, который исключён из Git.

```python
import os
from dotenv import load_dotenv
load_dotenv('.env')
api_key = os.getenv('OMDB_API_KEY', '').strip()
if not api_key:
    raise ValueError('Добавьте OMDB_API_KEY в локальный .env и повторите запуск.')
title = input('Введите название фильма: ').strip()
```

```python
response = requests.get('https://www.omdbapi.com/',
                        params={'t': title, 'apikey': api_key}, timeout=20)
if response.status_code != 200:
    raise RuntimeError(f'OMDb вернул HTTP {response.status_code}; проверьте ключ и доступность сервиса.')
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

```python
fields = {"Название": "Title", "Год": "Year", "Жанр": "Genre",
          "Режиссёр": "Director", "Рейтинг IMDb": "imdbRating"}
for label, key in fields.items():
    print(f"{label}: {json_data.get(key, chr(8212))}")
print(f"Актёры: {json_data.get("Actors", chr(8212))}")
```

Поиск `t=название` возвращает один фильм. Для списка совпадений используют `s`,
а для уточнения конкретного фильма можно передать год `y` или IMDb ID `i`.
Справочник: https://www.omdbapi.com/#parameters
