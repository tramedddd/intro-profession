```python
import json
import getpass
import requests
```

```python
title = input('Введите название фильма: ').strip()
api_key = getpass.getpass('Введите OMDb API key: ').strip()
```

```python
response = requests.get('https://www.omdbapi.com/',
                        params={'t': title, 'apikey': api_key}, timeout=20)
if response.status_code != 200:
    raise RuntimeError(f'OMDb вернул HTTP {response.status_code}; проверьте ключ и доступность сервиса.')
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

```python
fields = {"Название": "Title", "Год": "Year", "Жанр": "Genre",
          "Режиссёр": "Director", "Рейтинг IMDb": "imdbRating"}
for label, key in fields.items():
    print(f"{label}: {json_data.get(key, chr(8212))}")
print(f"Актёры: {json_data.get("Actors", chr(8212))}")
```
