Программа ищет фильм в OMDb, получает JSON и извлекает отдельные поля.
Для запуска нужен личный OMDb API key. Ячейки выполняются сверху вниз.

```python
import json
import getpass
import requests
```

Ключ читается из локального `.env`, который исключён из Git.

```python
import os
from dotenv import load_dotenv
load_dotenv('.env')
api_key = os.getenv('OMDB_API_KEY', '').strip()
if not api_key:
    raise ValueError('Добавьте OMDB_API_KEY в локальный .env и повторите запуск.')
title = input('Введите название фильма: ').strip()
```

После проверки кода добавлены таймаут, проверка входных данных и обработка ошибок.
Здесь выполнен обзор Codex; обзор GigaCode отдельно не проводился.
Не печатаем URL запроса или исключение requests: URL может содержать ключ.

```python
from movie_api import fetch_movie, movie_fields, MovieAPIError
try:
    json_data = fetch_movie(title, api_key)
    print(json.dumps(json_data, ensure_ascii=False, indent=2))
    for label, value in movie_fields(json_data).items():
        print(f'{label}: {value}')
except MovieAPIError as error:
    print(error)
```

Поиск `t=название` возвращает один фильм. Для списка совпадений используют `s`,
а для уточнения конкретного фильма можно передать год `y` или IMDb ID `i`.
